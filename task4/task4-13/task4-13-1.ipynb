```python
A = float(input("Введите A: "))
B = float(input("Введите B: "))
C = float(input("Введите C: "))
D = float(input("Введите D: "))
min_value = A

if B < min_value:
    min_value = B

if C < min_value:
    min_value = C

if D < min_value:
    min_value = D

print("Минимальное из введённых чисел:", min_value)
```
