```python
import pandas as pd
df = pd.read_csv('wild_boars.csv')

med = df.median(numeric_only=True)
result = pd.DataFrame({
    'Параметр': med.index,
    'Медианное_значение': med.values
})

result.to_csv('med_val.csv', index=False, encoding='utf-8')

print("Медианные значения:")
print(result)
print(f"\nСохранено в 'med_val.csv'")
```
