```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')
means = df.mean(numeric_only=True)

with open('avg_val.csv', 'w', encoding='utf-8') as f:
    f.write('Параметр,Среднее_значение\n')
    for param, value in means.items():
        f.write(f'{param},{value}\n')

print("Средние значения:")
print(means)
print("\nСохранено в 'avg_val.csv'")
```
