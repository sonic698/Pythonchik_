```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')

mal_tusk = df[df['gender'] == 'Male']['tusk_length_cm'].dropna()
fem_tusk = df[df['gender'] == 'Female']['tusk_length_cm'].dropna()

mal_cv = (mal_tusk.std() / mal_tusk.mean()) * 100
fem_cv = (fem_tusk.std() / fem_tusk.mean()) * 100

result = pd.DataFrame({
    'Пол': ['Самец', 'Самка'],
    'Коэффициент вариации длины клыков (%)': [round(mal_cv, 2), round(fem_cv, 2)]
})

result.to_csv('tusk_cv_by_gender.csv', index=False, encoding='utf-8')

print(result)
print("\nСохранено в 'tusk_cv_by_gender.csv'")
```
