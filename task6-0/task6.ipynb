```python
import pandas as pd
df = pd.read_csv('wild_boars.csv')
mal = df[df['gender'] == 'Male']      
fem = df[df['gender'] == 'Female'] 
mal_length = mal['length_cm']

mal_q1 = mal_length.quantile(0.5)
mal_q3 = mal_length.quantile(0.8)
mal_iqr = mal_q3 - mal_q1

fem_length = fem['length_cm']
fem_q1 = fem_length.quantile(0.5)
fem_q3 = fem_length.quantile(0.8)
fem_iqr = fem_q3 - fem_q1


result = pd.DataFrame({'Пол': ['Самец', 'Самка'], 'Q1 (50-й перцентиль)': [mal_q1, fem_q1], 'Q3 (80-й перцентиль)': [mal_q3, fem_q3], 'Межквартильный размах (IQR)': [mal_iqr, fem_iqr]})

result.to_csv('iqr_by_gender.csv', index=False)

print(result)
print("\nСохранено в 'iqr_by_gender.csv'")
```
