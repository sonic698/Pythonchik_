```python
import pandas as pd

df = pd.read_csv('wild_boars.csv')
num = df.select_dtypes(include=['number']).drop(columns=['boar_id'])

var = num.var()
std_dev = num.std()
cv = (std_dev / num.mean()) * 100

result = pd.DataFrame({
    'Параметр': var.index,
    'Дисперсия': var.values,
    'Стандартное_отклонение': std_dev.values,
    'Коэффициент_вариации_%': cv.values
})

result.to_csv('var_stat.csv', index=False, encoding='utf-8')
print(result)
```
